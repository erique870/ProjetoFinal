```python
merge_pt_en = []
wrong = [0, 995, 1436, 1555, 1914, 3924, 4699, 6222, 6757 ,7132]
offset = 0

for i in range(len(dataset_pt["train"])):
    if (i + offset) in wrong:
        offset += 1
    merge_pt_en.append((dataset_pt["train"][i], dataset_en["train"][i + offset]))
```

```python
len(merge_pt_en)
```

```python
import json

with open("data/pares_pt_en.json", "w", encoding="utf-8") as f:
    json.dump(merge_pt_en, f, ensure_ascii=False, indent=2)
```
